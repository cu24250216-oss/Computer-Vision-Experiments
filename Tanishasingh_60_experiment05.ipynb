**EXPERIMENT 05**

**Feature Extraction and Image Analysis using SIFT and HOG Descriptors**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
```

```python
uploaded = files.upload()

filename = list(uploaded.keys())[0]

image = cv2.imread(filename)

if image is None:
    print("Image not found!")
else:
    print("Image loaded successfully!")
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6, 6))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
sift = cv2.SIFT_create()

keypoints, descriptors = sift.detectAndCompute(gray, None)

print("Number of Keypoints:", len(keypoints))
print("Descriptor Shape:", descriptors.shape)
```

```python
sift_image = cv2.drawKeypoints(
    gray,
    keypoints,
    None,
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)

plt.figure(figsize=(8, 6))
plt.imshow(sift_image, cmap="gray")
plt.title("SIFT Keypoints")
plt.axis("off")
plt.show()
```

```python
from skimage.feature import hog

hog_features, hog_image = hog(
    gray,
    orientations=9,
    pixels_per_cell=(8, 8),
    cells_per_block=(2, 2),
    visualize=True
)

print("HOG Feature Length:", len(hog_features))
```

```python
plt.figure(figsize=(8, 6))

plt.imshow(hog_image, cmap="gray")
plt.title("HOG Feature Visualization")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(sift_image, cmap="gray")
plt.title("SIFT Keypoints")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(hog_image, cmap="gray")
plt.title("HOG Features")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
print("----- Feature Extraction Results -----")
print("SIFT Keypoints:", len(keypoints))
print("SIFT Descriptor Shape:", descriptors.shape)
print("HOG Feature Length:", len(hog_features))
print("Feature extraction completed successfully!")
```

**Questions**


**Q1. What is feature extraction, and why is it important in computer vision?**
**Ans: **Feature extraction finds important patterns like edges, corners, and shapes. It helps in image recognition and classification.

**Q2. Explain the working principle of SIFT.**
**Ans:** SIFT detects distinctive keypoints and generates descriptors that are resistant to scale and rotation changes.

**Q3. What are keypoints and feature descriptors?**
Ans: Keypoints are important points in an image. Descriptors are numerical values that describe the area around each keypoint.

**Q4. Explain HOG and its significance.**
**Ans:** HOG represents an image using gradient directions and strengths. It is useful for detecting object shapes and edges.

**Q5. Compare SIFT and HOG.**
**Ans: **SIFT is robust to scale and rotation and is useful for image matching. HOG focuses mainly on shape and edge information and is commonly used for object detection.

**Q6. Why is SIFT invariant to scale and rotation?**
**Ans**: SIFT detects features at multiple scales and assigns an orientation to each keypoint.

**Q8. Why is feature extraction done before classification?**
**Ans:** It extracts useful information from images, reducing unnecessary data and improving classification.

**Q9. Give advantages and limitations of handcrafted features.**
**Ans:** They are simple and require less data, but they may not perform well on complex images compared with deep-learning features.

**Q10. How are feature extraction techniques useful in image matching, face recognition, and object detection?**
**Ans:** They extract distinctive image features that help identify, compare, and recognize objects or faces.
