```python
!pip install ultralytics opencv-python matplotlib -q

import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

!wget -q https://ultralytics.com/images/bus.jpg -O bus.jpg

results = model("bus.jpg")

result_img = results[0].plot()

plt.figure(figsize=(10,7))
plt.imshow(cv2.cvtColor(result_img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()

print("Detected Objects:")

for box in results[0].boxes:
    cls = int(box.cls[0])
    conf = float(box.conf[0])
    print(model.names[cls], round(conf, 2))

cv2.imwrite("detected_objects.jpg", result_img)
print("Result saved as detected_objects.jpg")
```

**Questions**

1. Differentiate between image classification, object detection, and image segmentation.

Classification: Identifies the object class.
Object detection: Identifies objects and draws bounding boxes.
Segmentation: Classifies each pixel of an image.

2. Explain YOLO.

YOLO (You Only Look Once) detects objects in a single pass through a neural network. It predicts bounding boxes, class labels, and confidence scores

3. Compare YOLO, SSD, and Faster R-CNN.

Model	Speed	Accuracy
YOLO	Very Fast	High
SSD	Fast	High
Faster R-CNN	Slower	High

4. What is a bounding box?

A rectangular box around an object that shows its location and size.

5. What are confidence score and IoU?

Confidence score indicates prediction certainty. IoU measures overlap between predicted and actual bounding boxes.

6. Advantages of pre-trained models?

They save training time, require less data, and provide good accuracy.

7. Why are MS COCO and Pascal VOC used?

They contain labeled images and bounding-box annotations for training and evaluating object detection models.

8. Five applications of object detection.

Autonomous vehicles
Surveillance
Face detection
Medical imaging
Traffic monitoring

9. Challenges in object detection.

Poor lighting, occlusion, complex backgrounds, different object sizes, and overlapping objects.

10. How can object detection be improved?
